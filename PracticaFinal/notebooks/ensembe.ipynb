#### 3.2.1.3. Stack Ensemble

```julia
using Flux

KNNClassifier = @load KNNClassifier pkg=NearestNeighborModels verbosity=0
NeuralNetworkClassifier = @load NeuralNetworkClassifier pkg=MLJFlux verbosity=0
SVMClassifier = @load ProbabilisticSVC pkg=LIBSVM verbosity=0

mlp = NeuralNetworkClassifier(builder=MLJFlux.MLP(; hidden=(50,), σ=Flux.relu),
                                                    optimiser=Optimisers.Adam(0.01),
                                                    batch_size=8,
                                                    epochs=10,
                                                    rng=104)
knn = KNNClassifier(K = 10)
svm = SVMClassifier()

# Asumiendo que los modelos ya existían
stack = Stack(; metalearner = mlp,
                svm = svm,
                knn = knn,)
```
